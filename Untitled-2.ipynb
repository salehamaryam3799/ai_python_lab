```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set aesthetic styling
sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv('Crop_recommendation.csv')

# Create dashboard figure layout (2 rows x 2 columns)
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# -------------------------------------------------------------
# Graph 1: Feature Correlation Heatmap
# -------------------------------------------------------------
numeric_df = df.drop(columns=['label'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f", ax=axes[0, 0])
axes[0, 0].set_title('1. Feature Correlation Heatmap', fontsize=12, fontweight='bold')

# -------------------------------------------------------------
# Graph 2: Rainfall Requirements for Key Crops
# -------------------------------------------------------------
target_crops = ['rice', 'maize', 'chickpea', 'apple', 'coffee', 'cotton']
df_filtered = df[df['label'].isin(target_crops)]

avg_rain = df_filtered.groupby('label')['rainfall'].mean().reset_index()
sns.barplot(data=avg_rain, x='label', y='rainfall', ax=axes[0, 1], palette='crest')
axes[0, 1].set_title('2. Average Rainfall (mm) by Selected Crop', fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel('Crop Label')
axes[0, 1].set_ylabel('Rainfall (mm)')

# -------------------------------------------------------------
# Graph 3: Nitrogen (N) Distribution per Crop
# -------------------------------------------------------------
sns.boxplot(data=df_filtered, x='label', y='N', ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('3. Nitrogen (N) Distribution across Crops', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Crop Label')
axes[1, 0].set_ylabel('Nitrogen (kg/ha)')

# -------------------------------------------------------------
# Graph 4: Environmental Clusters (Temperature vs Humidity)
# -------------------------------------------------------------
sns.scatterplot(
    data=df_filtered, 
    x='temperature', 
    y='humidity', 
    hue='label', 
    style='label', 
    s=80, 
    ax=axes[1, 1]
)
axes[1, 1].set_title('4. Temperature vs. Humidity Profiles', fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel('Temperature (°C)')
axes[1, 1].set_ylabel('Humidity (%)')

# Final adjustments and file saving
plt.tight_layout()
plt.savefig('crop_recommendation_graphs.png', dpi=300)
plt.show()
```

```python
import csv

# -------------------------------------------------------------
# 1. User-Defined Rule-Based Decision Function
# -------------------------------------------------------------
def predict_crop_category(temp: float, humidity: float, rainfall: float) -> str:
    """
    Applies nested conditional logic on three environmental features
    (Temperature, Humidity, Rainfall) to generate a recommended crop category.
    """
    if rainfall > 180.0:
        if humidity > 70.0:
            return "Rice / High-Water Crop"
        else:
            return "Heavy Rainfall Crop"
    elif 100.0 <= rainfall <= 180.0:
        if temp > 25.0 and humidity > 60.0:
            return "Tropical Fruits (Banana/Papaya/Jute)"
        elif temp <= 25.0:
            return "Maize / Cotton / Plantation"
        else:
            return "Moderate Rainfall Crop"
    elif rainfall < 100.0:
        if temp < 22.0:
            return "Pulses / Legumes (Chickpea/Lentil)"
        elif humidity < 50.0:
            return "Arid / Low-Moisture Crop"
        else:
            return "Low-Water Crop"
    else:
        return "General Agricultural Crop"


def categorize_label(actual_label: str) -> str:
    """Maps specific dataset crop labels to generalized decision categories."""
    rice_group = ['rice']
    tropical_group = ['banana', 'papaya', 'jute', 'coconut', 'coffee', 'muskmelon']
    moderate_group = ['maize', 'cotton', 'apple', 'grapes', 'orange', 'pomegranate', 'watermelon']
    pulses_group = ['chickpea', 'lentil', 'blackgram', 'mungbean', 'mothbeans', 'pigeonpeas', 'kidneybeans']
    
    label_lower = actual_label.strip().lower()
    if label_lower in rice_group:
        return "Rice / High-Water Crop"
    elif label_lower in tropical_group:
        return "Tropical Fruits (Banana/Papaya/Jute)"
    elif label_lower in moderate_group:
        return "Maize / Cotton / Plantation"
    elif label_lower in pulses_group:
        return "Pulses / Legumes (Chickpea/Lentil)"
    else:
        return "General Agricultural Crop"

# -------------------------------------------------------------
# 2. Dataset Processing and Processing Loop
# -------------------------------------------------------------
file_path = 'Crop_recommendation.csv'

# Samples across different crop types for evaluation
sample_indices = [0, 100, 200, 300, 400, 600, 700, 800, 900, 1100]

evaluated_records = []
correct_matches = 0
total_evaluated = 0

with open(file_path, mode='r') as file:
    reader = list(csv.DictReader(file))
    
    for idx in sample_indices:
        if idx < len(reader):
            row = reader[idx]
            
            # Extract features (Temperature, Humidity, Rainfall)
            temp = float(row['temperature'])
            humidity = float(row['humidity'])
            rainfall = float(row['rainfall'])
            actual_crop = row['label']
            
            # Rule-based decision execution
            rule_decision = predict_crop_category(temp, humidity, rainfall)
            expected_category = categorize_label(actual_crop)
            
            # Evaluation check
            is_match = (rule_decision == expected_category)
            if is_match:
                correct_matches += 1
            total_evaluated += 1
            
            evaluated_records.append({
                'Index': idx,
                'Temp': temp,
                'Humidity': humidity,
                'Rainfall': rainfall,
                'Predicted_Category': rule_decision,
                'Actual_Label': actual_crop,
                'Expected_Category': expected_category,
                'Match': "MATCH" if is_match else "MISMATCH"
            })

# -------------------------------------------------------------
# 3. Output Display & Evaluation Results
# -------------------------------------------------------------
print("=" * 110)
print(f"{'Index':<6} | {'Temp(°C)':<8} | {'Hum(%)':<7} | {'Rain(mm)':<8} | {'Actual Crop':<12} | {'Predicted Category':<35} | {'Result':<8}")
print("=" * 110)

for rec in evaluated_records:
    print(f"{rec['Index']:<6} | {rec['Temp']:<8.1f} | {rec['Humidity']:<7.1f} | {rec['Rainfall']:<8.1f} | {rec['Actual_Label']:<12} | {rec['Predicted_Category']:<35} | {rec['Match']:<8}")

print("=" * 110)

# Accuracy Metric Calculation
accuracy_percentage = (correct_matches / total_evaluated) * 100.0 if total_evaluated > 0 else 0.0

print("\nSYSTEM EVALUATION SUMMARY:")
print(f"Total Evaluated Records : {total_evaluated}")
print(f"Correct Matches          : {correct_matches}")
print(f"Incorrect Mismatches     : {total_evaluated - correct_matches}")
print(f"Rule-Based Accuracy Rate : {accuracy_percentage:.2f}%")
```

```python
import csv

# Open and load dataset using Python's built-in csv module
records = []
with open('Crop_recommendation.csv', mode='r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        records.append(row)

# 1. Total Records and Features
print(f"Total Records: {len(records)}")
print(f"Total Features: {len(reader.fieldnames)}")
print("Column Names:", reader.fieldnames)

# 2. Extract Numerical Feature Statistics (e.g., Nitrogen 'N')
n_values = [float(row['N']) for row in records]
print(f"\nNitrogen (N) Min: {min(n_values)}")
print(f"Nitrogen (N) Max: {max(n_values)}")
print(f"Nitrogen (N) Average: {sum(n_values) / len(n_values):.2f}")

# 3. Unique Crop Categories
crop_types = sorted(list(set(row['label'] for row in records)))
print(f"\nTotal Crop Categories: {len(crop_types)}")
print("Crops:", crop_types)
```
