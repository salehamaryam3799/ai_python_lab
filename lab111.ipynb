```python
# ==========================================
# 1. Variables and Data Types
# ==========================================
sample_size = 100          # Integer: batch size
learning_rate = 0.01       # Float: hyperparameter
model_name = "Neural Net"  # String: model identifier
is_trained = False         # Boolean: status flag

# ==========================================
# 2. Lists and Operators
# ==========================================
feature_weights = [0.25, 0.85, 0.40, 0.95, 0.10]
feature_weights.append(0.60)  # Basic List Operation: Append

# ==========================================
# 3 & 4. Conditional Statements and Loops
# ==========================================
threshold = 0.5
high_weights = []

# 'for' loop with conditional filtering
for weight in feature_weights:
    if weight >= threshold and not is_trained:
        high_weights.append(weight)

# 'while' loop demonstration
epoch = 1
max_epochs = 3

# ==========================================
# 5. Functions & Error Handling
# ==========================================
def calculate_mean_accuracy(correct_predictions, total_samples):
    """Calculates accuracy percentage safely."""
    if total_samples <= 0:
        return 0.0  # Guard clause against zero-division
    return (correct_predictions / total_samples) * 100
```

```python
print("--- Data Types ---")
print(f"Model Name: {model_name} | Type: {type(model_name)}")
print(f"Learning Rate: {learning_rate} | Type: {type(learning_rate)}")

print("\n--- Operators & Lists ---")
print(f"Feature Weights List: {feature_weights}")
print(f"Weights >= {threshold}: {high_weights}")

print("\n--- Training Loop ---")
while epoch <= max_epochs:
    print(f"Training Epoch {epoch}/{max_epochs} complete.")
    epoch += 1

print("\n--- Function Output ---")
total = 150
correct = 135
acc = calculate_mean_accuracy(correct, total)
print(f"Total Samples: {total} | Correct Predictions: {correct}")
print(f"Model Accuracy: {acc:.2f}%")
```
