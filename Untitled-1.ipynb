```python
# ==========================================
# Complete, Runnable Python Demonstration
# ==========================================

def preprocess_dataset(raw_features, multiplier=1.5):
    """
    Demonstrates: Functions, Variables, Data Types, Arithmetic Operators,
    and List Operations.
    """
    threshold = 50.0  # Float variable
    scaled_features = []  # List initialization

    # For loop to iterate through the input list
    for feature in raw_features:
        # Arithmetic operator (*) and Type conversion
        adjusted_val = float(feature) * multiplier
        scaled_features.append(adjusted_val)  # Basic list operation

    return scaled_features, threshold


def evaluate_records(data_list):
    """
    Demonstrates: While Loops, Relational/Logical Operators, and Conditional Statements.
    """
    scaled_data, limit = preprocess_dataset(data_list)
    
    print(f"Original Input List: {data_list}")
    print(f"Processed Features : {scaled_data}\n")
    print("--- Individual Item Analysis ---")

    index = 0
    high_value_count = 0

    # While loop for conditional iteration
    while index < len(scaled_data):
        val = scaled_data[index]

        # Conditional logic using Relational (>) and Logical (and) operators
        if val > limit and val % 2 == 0:
            print(f"Record [{index}]: {val:<5.1f} -> STATUS: Above Threshold & Even")
            high_value_count += 1
        elif val > limit:
            print(f"Record [{index}]: {val:<5.1f} -> STATUS: Above Threshold")
            high_value_count += 1
        else:
            print(f"Record [{index}]: {val:<5.1f} -> STATUS: Normal")

        index += 1  # Increment loop counter to avoid infinite execution

    return high_value_count


# Main execution entry point
if __name__ == "__main__":
    raw_input_data = [20, 35, 40, 15, 60]
    total_high = evaluate_records(raw_input_data)
    
    print("\n--- Summary ---")
    print(f"Total entries passing threshold condition: {total_high}")
```
