```python
"""
Question 1: Core Python Fundamentals
Demonstrates: Variables, data types, arithmetic/relational/logical operators,
conditional statements, for/while loops, functions, and list operations.
"""

def classify_and_filter_data(sensor_readings, threshold):
    """
    Filters raw sensor data based on a numerical threshold.
    """
    valid_readings = []
    
    print(f"--- Processing {len(sensor_readings)} Sensor Data Points ---")
    
    # Loop 1: For loop traversing a list
    for index, value in enumerate(sensor_readings):
        # Relational and Logical Operators with type checking
        if isinstance(value, (int, float)) and value >= threshold:
            valid_readings.append(value)
            print(f"Index [{index}]: Reading {value} passes threshold ({threshold})")
        elif isinstance(value, (int, float)) and value < threshold:
            print(f"Index [{index}]: Reading {value} filtered out (Below threshold)")
        else:
            print(f"Index [{index}]: Invalid data type detected: {type(value)}")
            
    return valid_readings


def main():
    # 1. Variables and Data Types
    experiment_name = "Acoustic Target Detection"  # String
    sample_rate_khz = 48.0                        # Float
    active_status = True                          # Boolean
    raw_inputs = [12.5, 45.0, "ERR", 88.2, 5.1, 92.4, 31.0]  # List
    min_cutoff = 20.0                             # Float threshold

    print(f"Experiment: {experiment_name} | Rate: {sample_rate_khz} kHz | Active: {active_status}\n")

    # 2. Function Execution & List Manipulation
    filtered_data = classify_and_filter_data(raw_inputs, min_cutoff)

    # 3. Arithmetic Operations
    if filtered_data:
        total_sum = sum(filtered_data)
        average_val = total_sum / len(filtered_data)
        print(f"\nFiltered Results List: {filtered_data}")
        print(f"Total Sum: {total_sum:.2f} | Mean Signal Level: {average_val:.2f}")

    # 4. While Loop Demonstration
    print("\n--- Diagnostic Countdown ---")
    countdown = 3
    while countdown > 0:
        print(f"System ready in T-{countdown} seconds...")
        countdown -= 1
    print("Execution complete.")


if __name__ == "__main__":
    main()
```
